```python
#load csv file
import pandas as pd
df=pd.read_csv('DSA1080_Pandas_Student_Data.csv')
print(df.head())
print(df.shape)

#inspect the data
print(df.info())
print(df.describe())
print(df.isnull().sum())
```

```python
#selecting columns
print(df['Name'])
print(df[['Name','Score']])
print(df['Score'].mean())
```

```python
#selecting rows with loc 
print(df.loc[0])
print(df.loc[0,'Name'])
print(df.loc[0:4,['Name','Score']])
```

```python
#select with iloc
print(df.iloc[0:5])
```

```python
#filter table with conditions
high=df[df['Score']>70]
print(high[['Name','Score']])
city=df[df['City']=='Nairobi']
print(city.head())
```

```python
#combine conditions with and
selected=df[(df['Score']>70) & (df['StudyHours']>=10)]
print(selected)
```

```python
best=df.sort_values(by='Score', ascending=False)
print(best[['Name','Score']].head(5))
```

```python
#create a calculated column
df['Total'] = df['Score'] + df['StudyHours']
print(df[['Name','Score','StudyHours','Total']])
```

```python
#Missing values
df=pd.read_csv('DSA1080_Pandas_Student_Data.csv')
print(df.isnull().sum())
print(df[df['Score'].isnull()])
```

```python
cleaned=df.copy()
median_score=cleaned['Score'].median()
cleaned['Score']=cleaned['Score'].fillna(median_score)
cleaned['City']=cleaned['City'].fillna('Unknown')
```

```python
print(cleaned.duplicated().sum())
cleaned=cleaned.drop_duplicates()
print(cleaned['Score'].between(0, 100).all())
print(cleaned.shape)
```

```python
#summary statistics
print(cleaned['Score'].mean())
print(cleaned['Score'].median())
print(cleaned['Score'].std())
print(cleaned['Score'].min())
print(cleaned['Score'].max())
print(cleaned['Score'].count())
```

```python
#frequency counts
print(cleaned['City'].value_counts())
print(cleaned['Course'].value_counts())
print(cleaned['Course'].nunique())
```
